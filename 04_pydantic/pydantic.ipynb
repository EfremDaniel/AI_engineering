# Validate with pydantic

```python
from pydantic import BaseModel, ValidationError

# Student is a normal class but is also pydantic Basemodel as we inherit from it
class Student(BaseModel):
    name: str
    gender: str
    age: int

s1= Student(name="Breda", gender="F", age=8)
s1
```

```python
try:
    Student(name=2.453, gender=8, age=-8)
except ValidationError as err:
    print(err)
```

```python
from pydantic import Field
from typing import Literal

class Student(BaseModel):
    name: str
    gender: Literal["M", "F"]
    age: int = Field(gt= -1, lt= 125)

try: 
    Student(name = "Bibbi", gender= "female", age= -1)
except ValidationError as err:
    print(err)
```

```python
s2 = Student(name = "Bibbi", gender= "F", age= 1)
s2
```

# Serialization and deserialization

- serialization: convert objet > json
- deserialization: convert json > object

```python
s2.model_dump()
```

```python
import json

with open("housing.json", "r") as f:
    json_data = json.loads(f.read())

print(json_data)
```

```python
from pydantic import BaseModel
class House(BaseModel):
    address: str
    price: int
    area: float
    rooms: float
```

```python
House(**json_data[0])
```

```python
[House(**house_data) for house_data in json_data]
```
