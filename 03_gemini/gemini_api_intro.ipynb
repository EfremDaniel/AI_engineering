```python
from google import genai

client = genai.Client()

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Explain how AI works in a few words",
)

print(response.text)
```

```python
def ask_gemini(prompt, model="gemini-2.0-flash"):
    response = client.models.generate_content(
        model=model,
        contents=prompt
    )
    return response

response = ask_gemini("Give me 5 data engineering jokes, structured in short points")
print(response.text)

 
```

```python
from google.genai import types

prompt = ("Give me 5 some data engineering jokes, structure in short points")

response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents="prompt",
        config=types.GenerateContentConfig(
            thinking_config=types.ThinkingConfig(thinking_budget=8)
        ),
        )

print(response.text)

```

```python
response.usage_metadata
```

### System instruction

```python
system_instruction = """You are a joking robot called Ro Båt, which 
        will always answer with a programming joke.
        """

prompt = "What is the weather today?"

response = client.models.generate_content(
    model="gemini-2.0-flash",
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
    ),
    contents=prompt,
)

response.text
```

```python
print(f"{len(system_instruction.split()) = }")
print(f"{len(prompt.split()) = }")
print("plus some formatting overhead")

print(f"prompt token count {response.usage_metadata.prompt_token_count}")
```

### Temperature

- controls randomness of output -> 'creative'

```python
story_prompt = "write a 2 sentence story about a gray rabbit"

boring_story = client.models.generate_content(
    model="gemini-2.0-flash",
    contents=story_prompt,
    config=types.GenerateContentConfig(temperature=0.0),
)

print(boring_story.text)
```

### Multimodal input

input text and image

```python
text_input = "Describe this image shortly"
image_input = {"mime_type": "image/png", "data": open("bella.png", 'rb').read()}


response = client.models.generate_content(
    model="gemini-2.5-pro",
    contents=dict(
        parts=[dict(text = text_input), dict(inline_data = image_input)]
    )
)

print(response.text)
```
